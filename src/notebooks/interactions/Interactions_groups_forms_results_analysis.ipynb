```python
import polars as pl

# 1. Preparamos las columnas de interacciones que necesitamos
df_interact_subset = interactions_df.select([
    pl.col("id"),
    pl.col("WSDI"), 
    pl.col("WSDI_cat")
])

# 2. Unimos con el dataframe del parquet
# Usamos 'left' para mantener a todos los alumnos, 
# incluso a los del grupo control (que tendrán null en WSDI)
df_merged = processed_forms_data.join(df_interact_subset, on="id", how="left")

# 3.Filtrar para el gráfico
# Para el gráfico de dispersión del WSDI, solo nos interesan los que interactuaron
df_plot_ready = df_merged.filter(pl.col("WSDI").is_not_null())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# Convertimos a Pandas para Seaborn
plot_df = df_plot_ready.to_pandas()


# --- Configuración de Categorías ----

categories = [
    {"label": "Not_Relevant",        "range": (0.0, 0.5), "color": "#EF3636"},
    {"label": "Low",                 "range": (0.5, 1.5), "color": "#DD8F1A"},
    {"label": "Lower_Intermediate",  "range": (1.5, 2.0), "color": "#17CDB5"},
    {"label": "Upper_Intermediate",  "range": (2.0, 2.5), "color": "#417EDA"},
    {"label": "High",                "range": (2.5, 3.0), "color": "#20BD15"},
]


sns.set_theme(style="whitegrid")
fig, ax = plt.subplots(figsize=(11, 5))

# --- Fondo Coloreado ---
for cat in categories:
    ax.axvspan(cat["range"][0], cat["range"][1], alpha=0.12, color=cat["color"], zorder=0)
    x_mid = (cat["range"][0] + cat["range"][1]) / 2
    ax.text(x_mid, 1.05, cat["label"], color=cat["color"], fontsize=8, 
            fontweight="bold", ha="center", transform=ax.get_xaxis_transform())

# --- Scatter Plot ---
sns.scatterplot(
    data=plot_df,
    x="WSDI",
    y="puntuacion_tc_hake_gain", # Ajusta este nombre si en tu parquet es distinto
    hue="WSDI_cat",
    palette={c["label"]: c["color"] for c in categories},
    s=80, edgecolor="white", linewidth=0.5, ax=ax
)

# --- Estética Final ---
ax.set_title("Relación entre Calidad de Interacción y Ganancia de Aprendizaje", 
             fontsize=12, fontweight="bold", pad=20)
ax.set_ylim(-0.05, 1.05) # Hake gain normalizado
ax.set_xlim(0, 3)
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# 1. Preparar datos (Convertir a Pandas para Seaborn)
# Asumimos que df_merged tiene 'WSDI' y 'puntuacion_tc_hake_gain'
plot_df = processed_forms_data.to_pandas() 

# --- Category config (Mantenemos tu estilo original) ---
categories = [
    {"label": "Not Relevant",       "range": (0,   0.5),  "color": "#EF3636"},
    {"label": "Superficial",        "range": (0.5, 1.5),  "color": "#DD8F1A"},
    {"label": "Low Intermediate",   "range": (1.5, 2.0),  "color": "#17CDB5"},
    {"label": "High Intermediate",  "range": (2.0, 2.5),  "color": "#417EDA"},
    {"label": "Deep",               "range": (2.5, 3.0),  "color": "#20BD15"},
]

# --- Style ---
sns.set_theme(style="whitegrid")
fig, ax = plt.subplots(figsize=(12, 6))

# --- Background: Regiones coloreadas ---
for cat in categories:
    ax.axvspan(cat["range"][0], cat["range"][1], alpha=0.10, color=cat["color"], zorder=0)
    # Etiquetas de categoría en la parte superior
    x_mid = (cat["range"][0] + cat["range"][1]) / 2
    ax.text(x_mid, 1.02, cat["label"], color=cat["color"], fontsize=9, 
            fontweight="bold", ha="center", transform=ax.get_xaxis_transform())

# --- Resultados: Scatter Plot ---
sns.scatterplot(
    data=plot_df,
    x="WSDI",
    y="puntuacion_tc_hake_gain",
    hue="WSDI_cat", # Opcional: colorear puntos por categoría
    palette={c["label"]: c["color"] for c in categories},
    s=70,
    edgecolor="black",
    linewidth=0.6,
    alpha=0.8,
    ax=ax
)

# --- Línea de tendencia (Opcional pero muy útil para ver correlación) ---
sns.regplot(
    data=plot_df, x="WSDI", y="puntuacion_tc_hake_gain", 
    scatter=False, color="#575555", line_kws={"linestyle":"--", "alpha":0.5}, ax=ax
)

# --- Boundary lines ---
for cat in categories[1:]:
    ax.axvline(x=cat["range"][0], color="gray", linewidth=0.8, linestyle="--", alpha=0.3)

# --- Configuración de ejes y etiquetas ---
ax.set_title("Impacto de la Calidad de Interacción (WSDI) en el Aprendizaje (Hake Gain)", 
             fontsize=14, fontweight="bold", loc="left", pad=25)
ax.set_xlabel("WSDI Score (Calidad de la Conversación)", fontsize=11)
ax.set_ylabel("Ganancia de Hake (puntuacion_tc)", fontsize=11)

ax.set_xlim(0, 3)
ax.set_ylim(-0.1, 1.1) # La ganancia de Hake suele ir de 0 a 1
ax.set_xticks([0, 0.5, 1.0, 1.5, 2.0, 2.5, 3.0])

# --- Leyenda ---
patches = [mpatches.Patch(color=c["color"], label=c["label"]) for c in categories]
ax.legend(handles=patches, title="Calidad de Interacción", fontsize=9, loc="lower right")

plt.tight_layout()
plt.show()
```
