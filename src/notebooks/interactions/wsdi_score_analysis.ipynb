## Load necessary data

```python
script_path = os.getcwd()
project_path = os.path.join(script_path, '..', '..', '..')
raw_data_path = os.path.join(project_path, 'data', 'interactions', 'raw_data')
processed_data_path = os.path.join(project_path, 'data', 'interactions', 'processed_data')
```

```python
with open(os.path.join(raw_data_path, 'interactions_raw_data_20260422_112415.json'), 'r', encoding="utf-8") as f:
    interactions = json.load(f)
    
with open(os.path.join(processed_data_path, 'semantic_depth_data_20260422_203030.json'), 'r', encoding="utf-8") as f:
    semantic_depth = json.load(f)
```

```python
pl.read_parquet(
    fr"C:\Users\migue.MSI\Desktop\IBiDat\Papers\va-educational-impact\data\forms\processed_data\processed_pre_post_forms.parquet"
).filter(
    pl.col("id") == "RamiroMaeztu-pu9"
)
```

```python
interactions_with_semantic_depth = {}

for user_id, interaction in interactions.items():
    if user_id in semantic_depth:
        
        for i,elem in enumerate(interaction["chat_interactions"]):
            interactions_with_semantic_depth[f"{user_id}_{i}"] = {
                'interaction': elem["user_input"],
                'reasoning_semantic_depth_level': semantic_depth[user_id][i]["reasoning_semantic_depth_level"],
                'semantic_depth_level': semantic_depth[user_id][i]["semantic_depth_level"]
            }

interactions_with_semantic_depth_df = pl.DataFrame(
    data= [{'interaction_id': k, **v} for k, v in interactions_with_semantic_depth.items()]
)
```

```python
interactions_with_semantic_depth_df.sort("semantic_depth_level")
```

#### Análisis per semantic depth level

##### Level 0: Off-topic

```python
level_0_interactions = interactions_with_semantic_depth_df.filter(
    pl.col("semantic_depth_level") == 0
)

for interaction in level_0_interactions.iter_rows():
    interaction_id, interaction, reasoning_semantic_depth_level, semantic_depth_level = interaction
    print(f"Interaction ID: {interaction_id}")
    print(f"Interaction: {interaction}")
    print(f"Reasoning Semantic Depth Level: {reasoning_semantic_depth_level}")
    print(f"Semantic Depth Level: {semantic_depth_level}")
    print("\n")
```

Analizando por encima, existen varias cuestiones que son contextuales las cuales se categorizan como off-topic no siendo así. Por ejemplo "contestalo en un maximo de dos frases" claramente hace referencia a una pregunta anterior. Quizás habría que pasar el contexto de la conversación al evaluador para que pueda hacer una evaluación más precisa.

Por otro lado, un par de usuarios preguntan por Datatab que debe ser una herramienta que se menciona en el vídeo, como tal no es una pregunta relacionado con nada de estadística pero entiendo que se menciona en el vídeo así que no debería ser categorizado como "off-topic". 

##### Level 1: Superficial

```python
level_1_2_interactions = interactions_with_semantic_depth_df.filter(
    pl.col("semantic_depth_level").is_in([1, 2])
)

for interaction in level_1_2_interactions.iter_rows():
    interaction_id, interaction, reasoning_semantic_depth_level, semantic_depth_level = interaction
    print(f"Interaction ID: {interaction_id}")
    print(f"Interaction: {interaction}")
    print(f"Reasoning Semantic Depth Level: {reasoning_semantic_depth_level}")
    print(f"Semantic Depth Level: {semantic_depth_level}")
    print("\n")
```

Dentro de este grupo ahora mismo se inluirían lo que antes se categorizaba como nivel de profundidad 1 y 2. Sin embargo, habría que ver como definir exactamente ya que existen preguntas muy superficiales ("de que va el video"). 

Por otro lado, preguntas superficiales pero que al menos requieren de que el usuario haya prestado algo de atención al vídeo ("que es la desviacion tipica").

Finalmente, preguntas que se han categorizado dentro de este grupo pero que para mí deberían estar en un grupo superior: "esribeme un esquema con la difinicion de media,mediana, moda, estadistica descriptiva" o "diferencia de las dos ecuaciones de la desviacion estandar".

##### Level 2: Deep

```python
level_3_interactions = interactions_with_semantic_depth_df.filter(
    pl.col("semantic_depth_level") == 3
)

for interaction in level_3_interactions.iter_rows():
    interaction_id, interaction, reasoning_semantic_depth_level, semantic_depth_level = interaction
    print(f"Interaction ID: {interaction_id}")
    print(f"Interaction: {interaction}")
    print(f"Reasoning Semantic Depth Level: {reasoning_semantic_depth_level}")
    print(f"Semantic Depth Level: {semantic_depth_level}")
    print("\n")
```

Dentro de las preguntas, existen dos grupos diferenciados. Por un lado, usuarios que realizan preguntas elaboradas las cuales requieren de un razonamiento o un interés profundo en la materia. ("pero una pregunta, si hay 14 personas que van en coche otras 14 que van en bici, 5 que van en autobus y 5 que van andando. cual es la moda aqui?").

Y por otro lado, existen preguntas que es hacer un copia y pega de alguna pregunta del chat evaluador como podría ser ("Analice cómo la inclusión de la varianza junto a la desviación típica profundiza el entendimiento de la dispersión de una muestra en un entorno de investigación clínica.") la cual yo nunca categorizaría como pregunta profunda ya que se limita a hacer copiar y pegar, pasaría a ser una pregunta superficial.
