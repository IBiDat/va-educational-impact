#### Análisis per semantic depth level

##### Level 0: Off-topic

Por ejemplo "contestalo en un maximo de dos frases" claramente hace referencia a una pregunta anterior. Quizás habría que pasar el contexto de la conversación al evaluador para que pueda hacer una evaluación más precisa.

Dentro de las preguntas, existen dos grupos diferenciados. Por un lado, usuarios que realizan preguntas elaboradas las cuales requieren de un razonamiento o un interés profundo en la materia. Y por otro lado, existen preguntas que es hacer un copia y pega de alguna pregunta del chat evaluador como podría ser ("Analice cómo la inclusión de la varianza junto a la desviación típica profundiza el entendimiento de la dispersión de una muestra en un entorno de investigación clínica.") la cual yo nunca categorizaría como pregunta profunda ya que se limita a hacer copiar y pegar, pasaría a ser una pregunta superficial.

```python
import os
import json
import re

script_path = os.getcwd()
directory_path = os.path.abspath(os.path.join(script_path, '..', '..', '..'))
raw_data_path = os.path.join(directory_path, 'data', 'interactions', 'raw_data', 'interactions_raw_data.json')
template_path = os.path.join(directory_path, 'data', 'interactions', 'templates', 'semantic_depth_evaluation_template.md')
summary_path = os.path.join(directory_path, 'data', 'video_transcription', 'summary_video_transcription.txt')
questions_path = os.path.join(directory_path, 'data', 'interactions', 'raw_data', 'evaluation_questions.txt')
forms_dir = os.path.join(directory_path, 'data', 'forms', 'raw_data')


def retrieve_questions(source_text: str) -> list[str]:
    """
    Parses a structured text string to extract question numbers and their content.
    """
    # Regex pattern to find "QUESTION X:question: [The Question Text]"
    pattern = r"question:\s*(.*?)(?=\s*\[source|expected_answer:)"
    
    # re.findall returns only the captured group (the question text)
    questions = re.findall(pattern, source_text, re.DOTALL)
    
    # Clean up whitespace and remove empty strings
    return [q.strip() for q in questions if q.strip()]

def process_forms_questions(
    questions: list[str]
) -> list[str]:
    
    #Exclude 'Marca temporal', 'Puntuación', 'IDENTIFICADOR'
    questions = [
        q
        for q in questions
        if q not in ['Marca temporal', 'Puntuación', 'IDENTIFICADOR']
    ]
    
    #Remove [] elements
    pattern = r'\[.*?\]'
    clean_questions = [re.sub(pattern, '', pregunta).strip() for pregunta in questions]
    
    return clean_questions

def get_forms_questions(forms_dir: str) -> list[str]:
    pre_questions = pl.read_csv(
        os.path.join(forms_dir, "respuestas_forms_pre.csv")
    ).columns
    
    post_questions = pl.read_csv(
        os.path.join(forms_dir, "respuestas_forms_post.csv")
    ).columns
    
    clean_pre_questions = process_forms_questions(
        questions=pre_questions
    )
    clean_post_questions = process_forms_questions(
        questions=post_questions
    )
    
    return clean_pre_questions + clean_post_questions

def load_prompt():
    #Load prompt template
    with open(template_path, "r", encoding="utf-8") as f:
        original_prompt_template = f.read()
    
    #Load video transcription summary
    with open(summary_path, "r", encoding="utf-8") as f:
        video_summary = f.read()
    
    #Load evaluation questions
    with open(questions_path, "r", encoding="utf-8") as f:
        evaluation_questions = f.read()
    
    #### PROCESS QUESTIONS ####
    #Retrieve questions list from the evaluation questions text
    questions_list = retrieve_questions(evaluation_questions)
    
    #Create a str containing all the questions separated by new lines
    questions = "\n".join([f"Question {i+1}: {q}" for i, q in enumerate(questions_list)])
    ############################
    
    #Apply string formatting to the prompt template
    prompt_template = original_prompt_template.format(
        summary=video_summary,
        questions=questions,
        conversation_context="__CONVERSATION_CONTEXT__",
        user_interaction="__USER_INTERACTION__"
    )
    
    return prompt_template


def generate_semantic_depth_index(raw_data):

    semantic_depth_data = {}
    
    prompt_template = load_prompt()

    data_ids = list(raw_data.keys())
    total_ids = len(data_ids)

    for idx, data_id in enumerate(data_ids, start=1):
        user_interactions = raw_data[data_id]['chat_interactions']
        print("user_interactions")
        print(user_interactions)

        if user_interactions:
            for i, interaction in enumerate(user_interactions, start=1):
                context = user_interactions[:i-1]
                
                prompt = prompt_template.replace("__CONVERSATION_CONTEXT__", str(context))
                prompt = prompt.replace("__USER_INTERACTION__", str(interaction))
                
                print(prompt)
                print("\n")
                
                if len(user_interactions) > 2:
                    with open(os.path.join(directory_path,  f'semantic_depth_prompt_example.txt'), 'w', encoding='utf-8') as f:
                        f.write(prompt)
        else:
            pass
    
        print("-"*25)
        
    return semantic_depth_data


with open(raw_data_path, "r", encoding="utf-8") as f:
    raw_data = json.load(f)

generate_semantic_depth_index(
    raw_data=raw_data
)

get_forms_questions(
    forms_dir=forms_dir
)
```
