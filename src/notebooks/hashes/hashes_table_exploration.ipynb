```python
import os
import polars as pl
```

```python
script_path = os.getcwd()
project_path = os.path.join(script_path, '..', '..', '..')
hashes_table_path = os.path.join(project_path, 'data', 'hashes', 'processed_data', 'hashes_groups.csv')

hashes_table = pl.read_csv(hashes_table_path, separator=';')
```

## Hashes-Groups Table by Centro

```python
for centro in hashes_table['centro'].unique():
    print(f'\nHashes-Groups Table for Centro {centro}')
    with pl.Config(tbl_rows=-1):
        display(hashes_table.filter(pl.col('centro') == centro).select(['id', 'grupo']))
```
