# Análisis de los tiempos anómalos sobre la duración en los cuestionarios

```python
# ─── CONFIGURACIÓN ───────────────────────────────────────────────────────────
COL_TS_PRE   = "marca temporal_pre"
COL_TS_POST  = "marca temporal_post"
COL_CENTRO   = "centro"
COL_GRUPO    = "grupo"
COL_ID       = "alumno_id"

UMBRAL_MINIMO_SEG = 60 * 10
UMBRAL_MAXIMO_SEG = 60 * 20

df = processed_forms_data

df = df.with_columns([
    pl.col(COL_TS_PRE).str.to_datetime(),
    pl.col(COL_TS_POST).str.to_datetime(),
])

df = df.filter(pl.col(COL_CENTRO).is_not_null() & pl.col(COL_GRUPO).is_not_null())

# ─── 1. DETECCIÓN DE TIMESTAMPS ATÍPICOS POR SESIÓN ─────────────────────────
# Trabajamos sobre el timestamp en segundos desde el inicio de la sesión.
# Esto es solo para tener una escala numérica común; no implica que sea una
# duración real. La detección de outliers se hace sobre esa magnitud.

def marca_atipicos(pdf, col_ts, sufijo):
    """
    Marca timestamps atípicamente tempranos/tardíos respecto a su sesión
    usando el criterio clásico de Tukey (1.5 · IQR).
    """
    pdf[f"ts_seg_{sufijo}"]   = np.nan
    pdf[f"temprano_{sufijo}"] = False
    pdf[f"tardio_{sufijo}"]   = False

    for (c, g), sub in pdf.groupby([COL_CENTRO, COL_GRUPO]):
        ts = pd.to_datetime(sub[col_ts])
        ts_seg = (ts - ts.min()).dt.total_seconds().to_numpy(dtype=float)
        pdf.loc[sub.index, f"ts_seg_{sufijo}"] = ts_seg

        q1, q3 = np.percentile(ts_seg, [25, 75])
        iqr = q3 - q1
        low  = q1 - 1 * iqr
        high = q3 + 1.5 * iqr

        pdf.loc[sub.index, f"temprano_{sufijo}"] = ts_seg < low
        pdf.loc[sub.index, f"tardio_{sufijo}"]   = ts_seg > high


pdf = df.to_pandas()
marca_atipicos(pdf, COL_TS_PRE,  "pre")
marca_atipicos(pdf, COL_TS_POST, "post")

pdf["sospechoso_pre"]  = pdf["temprano_pre"]  | pdf["tardio_pre"]
pdf["sospechoso_post"] = pdf["temprano_post"] | pdf["tardio_post"]
pdf["sospechoso"]      = pdf["sospechoso_pre"] | pdf["sospechoso_post"]

df = pl.from_pandas(pdf)


# ─── 2. RESUMEN POR SESIÓN ──────────────────────────────────────────────────
resumen = (df.group_by([COL_CENTRO, COL_GRUPO])
             .agg(
                 pl.len().alias("n"),
                 pl.col("temprano_pre").sum().alias("temprano_pre"),
                 pl.col("tardio_pre").sum().alias("tardio_pre"),
                 pl.col("temprano_post").sum().alias("temprano_post"),
                 pl.col("tardio_post").sum().alias("tardio_post"),
                 pl.col("sospechoso").sum().alias("total_sosp"),
             )
             .sort([COL_CENTRO, COL_GRUPO]))

n = df["sospechoso"].sum()
print(f"Sospechosos: {n}/{df.height} ({100*n/df.height:.1f}%)\n")
resumen
```

```python
# ─── 3. STRIP PLOT: TIMESTAMPS REALES POR SESIÓN ────────────────────────────
# Este es el gráfico más informativo: ves cada alumno como un punto en su
# sesión real. Los primeros y últimos quedan visualmente identificados.

pdf = df.to_pandas().copy()
pdf["sesion"] = pdf[COL_CENTRO] + " · " + pdf[COL_GRUPO]
sesiones = sorted(pdf["sesion"].unique())

fig, axes = plt.subplots(2, 1, figsize=(13, 9))

for ax, col_ts, test in [(axes[0], COL_TS_PRE, "pre"),
                         (axes[1], COL_TS_POST, "post")]:
    for i, s in enumerate(sesiones):
        sub = pdf[pdf["sesion"] == s]
        ts = pd.to_datetime(sub[col_ts])
        # Usamos solo la hora del día para que las 8 sesiones se vean alineadas
        hora_decimal = ts.dt.hour + ts.dt.minute/60 + ts.dt.second/3600
        sosp = sub[f"sospechoso_{test}"]

        ax.scatter(hora_decimal[~sosp], [i]*sum(~sosp), alpha=0.5, s=30)
        ax.scatter(hora_decimal[sosp],  [i]*sum(sosp),  color="red", s=60,
                   zorder=3, label="atípico" if i == 0 else None)

    ax.set_yticks(range(len(sesiones)))
    ax.set_yticklabels(sesiones)
    ax.set_xlabel("Hora del día (h)")
    ax.set_title(f"Timestamps de envío — test {test.upper()}")
    ax.legend(loc="upper right")
    ax.grid(axis="x", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# ─── 4. BOXPLOT DE TIMESTAMPS POR SESIÓN ────────────────────────────────────
# Igual que antes pero sobre el propio timestamp (en segundos desde el inicio
# de la sesión, solo como escala). Los rojos son los marcados como atípicos.

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

for ax, test in zip(axes, ["pre", "post"]):
    col = f"ts_seg_{test}"
    datos = [pdf.loc[pdf["sesion"] == s, col].to_numpy() / 60 for s in sesiones]
    ax.boxplot(datos, labels=sesiones, showfliers=True)
    for i, s in enumerate(sesiones, start=1):
        sub = pdf[(pdf["sesion"] == s) & pdf[f"sospechoso_{test}"]]
        ax.scatter([i]*len(sub), sub[col]/60, color="red", s=40, zorder=3,
                   label="atípico" if i == 1 else None)
    ax.set_ylabel("Minutos desde el primer envío de la sesión")
    ax.set_title(f"Test {test.upper()}")
    ax.tick_params(axis="x", rotation=30)
    ax.legend()

plt.tight_layout()
plt.show()
```
